```python
from pyspark.sql import SparkSession 
from pyspark.sql.types import *

spark=SparkSession.builder.appName("Spark application").getOrCreate()

data=[
    (1,"Rahul","IT",50000),
    (2,"Rani","HR",70000),
    (3,"Arjun","Finance",60000),
    (4,"Priya","IT",55000),
    (5,"Ravi","IT",65000)
]

schema=StructType([
    StructField("id",IntegerType(),True),
    StructField("name",StringType(),True),
    StructField("department",StringType(),True),
    StructField("salary",IntegerType(),True)
])

df=spark.createDataFrame(data,schema)

df.show()
```

```python
df.printSchema()
```

```python
data2=[
    (1, "  Rahul  ", "rahul@gmail.com", "IT,Hyderabad", "2026-01-15", "2026-01-15 09:30:00"),
    (2, "PRIYA", "PRIYA@GMAIL.COM", "HR,Bangalore", "2025-06-20", "2025-06-20 10:45:00"),
    (3, " arjun ", None, "Finance,Chennai", "2024-03-10", "2024-03-10 08:15:00"),
    (4, "Sneha123", " sneha@gmail.com ", "IT,Delhi", None, None),
    (5, "John-Doe", "john@gmail.com", "IT,Mumbai", "2026-08-01", "2026-08-01 14:20:00")
]

schema2=StructType([
    StructField("id", IntegerType(), True),
    StructField("name", StringType(), True),
    StructField("email", StringType(), True),
    StructField("department_location", StringType(), True),
    StructField("joining_date",StringType(), True),
    StructField("last_login", StringType(), True)
])

df2=(
    spark.createDataFrame(data2,schema2)
    .withColumn("joining_date",to_date(col("joining_date"),"yyyy-MM-dd"))
    .withColumn("last_login",to_timestamp(col("last_login"),"yyyy-MM-dd HH:mm:ss"))
    )
df2.show()

df2.printSchema()
```

### Aggregations

count() <br>
avg() <br>
sum() <br>
min() <br>
max() <br>

```python
df3.groupBy("department").count().alias("count_by_dept").show()
```

```python
df3.groupBy("department").avg("salary").show()
```

```python
df3.groupBy("department").max("salary").show()
```

```python
df3.groupBy("department").agg(
    avg("salary").alias("avg_salary")
).show()
```

```python
df3.groupBy("department").agg(
    avg("salary").alias("avg_salary"),
    max("salary").alias("max_salary"),
    min("salary").alias("min_salary")
).show()
```

count(*)- counts everything in the dataset along with nulls  <br>
count("salary")- counts non null salary values

### having/filtering aggrgation results

```python
df3.groupBy("department").agg(
    avg("salary").alias("average_salary")
    ).filter(col("average_salary") > 50000).show()
```
