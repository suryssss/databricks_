```python
from pyspark.sql import functions as F

orders = (
    spark.range(0, 10_000_000)
    .withColumn(
        "customer_id",
        (F.col("id") % 100_000).cast("long")
    )
    .withColumn(
        "amount",
        (F.rand() * 1000).cast("double")
    )
    .drop("id")
)

orders.show(5)
```

```python
customers = (
    spark.range(0, 100_000)
    .withColumnRenamed("id", "customer_id")
    .withColumn(
        "customer_name",
        F.concat(F.lit("Customer_"), F.col("customer_id"))
    )
)

customers.show(5)
```

```python
normal_join = orders.join(
    customers,
    "customer_id"
)
```

```python
normal_join.explain("formatted")
```

```python
from pyspark.sql.functions import broadcast

broadcast_join = orders.join(
    broadcast(customers),
    "customer_id"
)
```

```python
broadcast_join.explain("formatted")
```
