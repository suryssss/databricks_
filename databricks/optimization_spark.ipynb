```python
%sql
    CREATE OR REPLACE TABLE demo_customers AS
SELECT
    id AS customer_id,
    concat('Customer_', id) AS customer_name,
    CASE
        WHEN id % 5 = 0 THEN 'Hyderabad'
        WHEN id % 5 = 1 THEN 'Bangalore'
        WHEN id % 5 = 2 THEN 'Chennai'
        WHEN id % 5 = 3 THEN 'Mumbai'
        ELSE 'Delhi'
    END AS city
FROM range(1, 100001);
```

```python
%sql
CREATE OR REPLACE TABLE demo_orders AS
SELECT
    id AS order_id,
    CAST((id % 100000) + 1 AS BIGINT) AS customer_id,
    CAST((id % 1000) * 10.5 AS DOUBLE) AS amount,
    date_add('2024-01-01', CAST(id % 365 AS INT)) AS order_date
FROM range(1, 10000000);
```

```python
%sql
SELECT
    o.order_id,
    o.customer_id,
    o.amount,
    c.customer_name,
    c.city
FROM demo_orders o
JOIN demo_customers c
    ON o.customer_id = c.customer_id;
```

```python
%sql
EXPLAIN FORMATTED
SELECT
    o.order_id,
    o.customer_id,
    o.amount,
    c.customer_name,
    c.city
FROM demo_orders o
JOIN demo_customers c
    ON o.customer_id = c.customer_id;
```

```python
%sql
create or replace table demo_large_customers as 
select
    id as customer_id,
    repeat('some_large_customer_information',100) as customer_info

from range(1,5000000)
```

```python
%sql
select /*+ BROADCASTJOIN */
    o.order_id,
    o.amount,
    lc.customer_info
from demo_orders o
join demo_large_customers lc
on o.customer_id=lc.customer_id
```
