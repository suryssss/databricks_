```python
from pyspark.sql import functions as F

df = spark.range(0, 10_000_000)

df = df.withColumn(
    "category",
    F.when(F.col("id") % 3 == 0, "A")
     .when(F.col("id") % 3 == 1, "B")
     .otherwise("C")
)

df = df.withColumn(
    "amount",
    (F.col("id") * 10) % 1000
)

display(df.limit(10))
```

```python
'''
df.cache()=df.persist(StorageLevel.MEMORY_AND_DISK_DESER)
'''
```

```python
res=(
    df
    .filter(F.col("amount")>500)
    .groupBy("category")
    .agg(
        F.sum("amount").alias("total_amount"),
        F.avg("amount").alias("avg_amount")
    ) 
     )
```

```python
res.show()
```
