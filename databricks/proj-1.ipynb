```python
customers_df=spark.read.format("csv").option("inferSchema",True).option("header",True).load("/Volumes/workspace/default/proj1/customers (1).csv")

orders_df=spark.read.format("csv").option("inferSchema",True).option("header",True).load("/Volumes/workspace/default/proj1/orders (1).csv")

products_df=spark.read.format("csv").option("inferSchema",True).option("header",True).load("/Volumes/workspace/default/proj1/products (1).csv")
```

```python
from pyspark.sql.functions import col
customers_df.select(col("customer_id").alias("c_id"),col("age"),col("city")).show()
```

```python
orders_df.filter(col("customer_id")=="C999").show()
```

```python
orders_df=orders_df.withColumn("total_amount",col("quantity")*col("price"))
```

```python
orders_df.show()
```

```python
orders_df=orders_df.withColumn("discount_amount",col("total_amount")*0.1)
```

```python
orders_df.withColumn("final_amount",col("total_amount")-col("discount_amount"))
```

```python
orders_df.show()
```

```python
from pyspark.sql.functions import when
customers_df=customers_df.withColumn(
    "customer_type",
    when(col("age")>=30,"Senior")
    .otherwise("young")                        
                            )
```

```python
customers_df.show()
```

```python
orders_df=orders_df.withColumn(
    "order_value_category",
    when(col("total_amount")>=50000,"High").
    when(col("total_amount")>=10000,"Medium").
    otherwise("Low")
)
```

```python
orders_df.show()
```

```python
from pyspark.sql.functions import upper
customers_df=customers_df.withColumn("name",upper(col("name")))

customers_df.show()
```

```python
from pyspark.sql.functions import sum , count , avg
orders_df.agg(count(col("order_id")).alias("total_orders")).display()
```

```python
orders_df.agg(sum(col("quantity")).alias("total_quantity")).show()
```

```python
orders_df.agg(sum(col("total_amount")).alias("total_revenue")).show()
```

```python
orders_df.agg(avg(col("total_amount")).alias("avg_revenue")).show()
```

```python
from pyspark.sql.functions import max, min, col

orders_df.agg(
    max(col("total_amount")).alias("max_revenue"),
    min(col("total_amount")).alias("min_revenue")
).show()
```

```python
orders_df.groupBy("category").agg(sum(col("total_amount")).alias("total_revenue")).show()
```

```python
orders_df.groupby("category").agg(count(col("order_id")).alias("no_of_orders")).show()
```

```python
orders_df.groupBy("category").agg(
    sum(col("total_amount")).alias("total_revenue"),
    count(col("order_id")).alias("no_of_orders"),
    avg(col("total_amount")).alias("avg_revenue"),
    max(col("total_amount")).alias("max_revenue"),
    min(col("total_amount")).alias("min_revenue")
).show()
```
