```python
from pyspark.sql import SparkSession
spark=SparkSession.builder.appName("Spark Data").getOrCreate()
```

```python
df=spark.read.format("csv").option("inferSchema",True).option("header",True).load("/Volumes/workspace/default/e_commerce/raw_data/customers.csv")
```

```python
df.display()
```

```python
from pyspark.sql.types import StructType,StructField, StringType, IntegerType
schema = StructType([ 
    StructField("customer_id", IntegerType(), True), 
    StructField("name", StringType(), True), 
    StructField("gender", StringType(), True), 
    StructField("email", StringType(), True)
    ])
df
```

```python
df_explict= spark.read.format("csv").option("header",True).schema(schema).load("/Volumes/workspace/default/e_commerce/raw_data/customers.csv")
```

```python
df_explict.display()
```

```python
from pyspark.sql.functions import col , count
df.filter(col("customer_id").isNull()).count()
```

```python
df.filter(col("email").isNull()).show()
```

```python
from pyspark.sql.functions import col,sum

df.select([
    sum(col(c).isNull().cast("int")).alias(c)
    for c in df.columns
]).show()
```

```python
df=df.dropna(subset=["email"])
```

```python
df.filter(col("email").isNull()).count()
```

```python

df.select([
    sum(col(c).isNull().cast("int")).alias(c)
    for c in df.columns
]).show()
```

```python
df.groupBy(df.columns).count().filter("count>1").show()
```

```python
df=df.dropDuplicates()
```

```python
df.groupBy(df.columns).count().filter("count>1").show()
```

```python
from pyspark.sql.functions import col,trim,initcap

df.withColumn(
    "name",trim(initcap(col("name")))
)
```

```python
df.display()
```

```python
from pyspark.sql.functions import lit
df=df.withColumn(
    "active",lit(True))
```

```python
df.display()
```

```python
df=df.withColumnRenamed("customer_id","c_id")
```

```python
df.display()
```
